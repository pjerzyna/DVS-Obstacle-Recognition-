Sprawdzenie parametrow mojego nagrania za pomoca komendy od Metavision SDK:

`$ metavision_file_info -i data_in/hello.raw`

Uruchomienie ramki po 40ms i pokazanie zarysu postaci:

`$ python3 main.py`

Wymagane jest zainstalowanie konfiguracja pakietow od Prophesee, to są kroki które wykonałem, uwaga pracuję w środowisku conda:

1. Instalacja bibliotek systemowych

`$ sudo apt-get update` 

`$ sudo apt-get install -y libusb-1.0-0-dev cmake build-essential libprotobuf-dev protobuf-compiler libv4l-dev libopencv-dev git`

`$ sudo apt-get install -y libboost-all-dev`

`$ sudo apt-get install -y pybind11-dev python3-pybind11`

`$ sudo apt-get install -y libgl1-mesa-dev libglu1-mesa-dev libegl1-mesa-dev freeglut3-dev libglew-dev libglfw3-dev`

2. Pobranie oficjalnego kodu OpenEB (Metavision)

`git clone https://github.com/prophesee-ai/openeb.git`

`cd openeb`

3. Pozwalam Condzie (środowisko DVS) widzieć pakiety - instalacja zależności

`$ conda activate DVS`

`$ pip install numpy opencv-python-headless protobuf pybind11 h5py`

4. Kompilacja, instalacja i dodanie ścieżek od konfiguracji:

`mkdir build && cd build`

`cmake .. -DBUILD_TESTING=OFF -DPython3_EXECUTABLE=$(which python)`

`make -j$(nproc)`

`sudo make install`

`$ sudo sh -c "echo '/usr/local/lib' > /etc/ld.so.conf.d/metavision.conf"`

`$ sudo ldconfig` <-- odświeżenie pamięci podręcznej linkera
